```python
def empty_grid():
  # The naive [0] * 3 created multiple aliased lists!
  return [[0 for _ in range(3)] for _ in range(3)]
g = empty_grid()
print(g)
```

```python
def print_grid(g):
  symbols = ' OX'
  for r in g:
    print(''.join([symbols[i] for i in r]))
```

```python
def valid_moves(g):
  moves = []
  for y in range(len(g)):
    for x in range(len(g[y])):
      if g[y][x] == 0:
        moves.append((y, x))
  return moves

ms = valid_moves(g)
ms
```

```python
def map_move(n):
  # This is the numeric keypad
  return 2 - (n - 1) // 3, (n - 1) % 3
  # This is the telephone keypad
  # return (n - 1) // 3, (n - 1) % 3

def check_mapping():
  for i in range(1, 10):
    print(f"{i} --> {map_move(i)}")

check_mapping()
```

```python
def get_move(moves):
  while len(moves) > 0:
    value = input("Make your move:")
    if value.isdigit() and len(value) == 1:
      move = map_move(int(value))
      if move in moves:
        return move
      else:
        print("That move is already taken!")
    else:
      print("Please enter 1-9!")
  return None

get_move(ms)
```

```python
def game_state(g):
  # Check for a winner
  # Are all of a row complete?
  for r in g:
    if r[0] > 0 and (r[0] == r[1] == r[2]):
      return r[0]
  # Are all of a column complete?
  for c in range(len(g[0])):
    if g[0][c] > 0 and (g[0][c] == g[1][c] == g[2][c]):
      return g[0][c]
  # Is a diagonal complete?
  if g[0][0] > 0 and g[0][0] == g[1][1] == g[2][2]:
    return g[0][0]
  if g[0][2] > 0 and g[0][2] == g[1][1] == g[2][0]:
    return g[0][2]
  if len(valid_moves(g)) == 0:
    return -1
  return 0

game_state(g)
```

```python
g = empty_grid()
p = 0
while game_state(g) == 0:
  print_grid(g)
  print(f"Player {p + 1}:")
  r, c = get_move(valid_moves(g))
  g[r][c] = p + 1
  p = (p + 1) % 2
```

```python
if game_state(g) < 0:
  print("Tie!")
else:
  print(f"Winner is Player {game_state(g)}!")
```
