```python
# CODE HERE
base_url = 'http://quotes.toscrape.com'
url = '/'
unique_authors = set()
while url != None:
  if url.startswith('/'):
    url = base_url + url
  print(url)
  html = requests.get(url).text
  soup = bs4.BeautifulSoup(html)
  for i in soup.select(".author"):
    unique_authors.add(i.text)
  next_url = soup.select(".next")
  if len(next_url) == 0:
    url = None
  else:
    url = next_url[0].select('a')[0]['href']
print(unique_authors)
```
